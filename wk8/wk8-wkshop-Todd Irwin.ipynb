```python
# Which store has the largest floor space?
df = pd.read_csv('wk8-stores.csv')
df.nlargest(1,'size_sqft')
```

```python
# What is the distribution of customers by state?
df = pd.read_csv('wk8-customers.csv')
df['state'] = df['state'].str.replace('California', 'CA')
df['state'] = df['state'].str.replace('North Carolina', 'NC')
df['state'] = df['state'].str.replace('Texas', 'TX')
df['state'] = df['state'].str.replace('Washington', 'WA')
df['state'] = df['state'].str.replace('Oregon', 'OR')
df['state'] = df['state'].str.replace('Maryland', 'MD')
df['state'].value_counts()
```
